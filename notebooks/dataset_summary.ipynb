```python
import pandas as pd
import numpy as np
from pathlib import Path
import os
```

```python
def analyze_dataset(filepath):
    """Extract key statistics from a dataset."""
    df = pd.read_csv(filepath)

    # File size in MB
    size_mb = os.path.getsize(filepath) / (1024 * 1024)

    # Basic counts
    n_samples = len(df)
    n_features = len(df.columns)

    # Identify feature types by prefix
    cat_features = [c for c in df.columns if c.startswith("cat_")]
    num_features = [c for c in df.columns if c.startswith("num_")]
    date_features = [c for c in df.columns if c.startswith("dat_")]
    cls_labels = [c for c in df.columns if c.startswith("cls_")]
    reg_labels = [c for c in df.columns if c.startswith("reg_")]
    id_features = [c for c in df.columns if c.startswith("id_")]

    # Label info
    num_labels = len(cls_labels) + len(reg_labels)
    num_unique_labels = 0
    if cls_labels:
        num_unique_labels = df[cls_labels[0]].nunique()

    # NaN percentage
    total_nan_pct = (df.isna().sum().sum() / (n_samples * n_features)) * 100

    # Memory usage
    memory_mb = df.memory_usage(deep=True).sum() / (1024 * 1024)

    return {
        "name": filepath.stem,
        "n_samples": n_samples,
        "n_features": n_features,
        "n_categorical": len(cat_features),
        "n_numerical": len(num_features),
        "n_date": len(date_features),
        "n_id": len(id_features),
        "cat_labels": len(cls_labels),
        "num_labels": len(reg_labels),
        "total_labels": num_labels,
        "n_unique_labels": num_unique_labels,
        "file_size_mb": round(size_mb, 2),
        "memory_mb": round(memory_mb, 2),
        "nan_pct": round(total_nan_pct, 2),
    }
```

```python
# Iterate through all datasets
data_dir = Path("../data")
csv_files = sorted(data_dir.glob("*.csv"))

results = []
for filepath in csv_files:
    print(f"Processing {filepath.name}...")
    try:
        stats = analyze_dataset(filepath)
        results.append(stats)
    except Exception as e:
        print(f"Error processing {filepath.name}: {e}")

# Create summary dataframe
summary_df = pd.DataFrame(results)
print(f"\nProcessed {len(summary_df)} datasets")
```

```python
# Display summary
summary_df.sort_values(by="n_samples", ascending=False, inplace=True)
summary_df
```

```python
# Check if any dataset has a total_labels different from 1
mismatched_labels = summary_df[summary_df["total_labels"] != 1]
if not mismatched_labels.empty:
    print("\nDatasets with total_labels != 1:")
    print(mismatched_labels[["name", "total_labels"]])
```

```python
# List duplicates by n_samples
duplicates = summary_df[summary_df.duplicated(subset=["n_samples"], keep=False)]
if not duplicates.empty:
    print("\nDatasets with duplicate number of samples:")
    print(duplicates[["name", "n_samples"]].sort_values(by="n_samples"))
```
